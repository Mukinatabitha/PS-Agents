```python
highway_graph = {
    "Nairobi": [("Ruiru", 20), ("Naivasha", 90)],
    "Ruiru": [("Thika", 25)],
    "Thika": [("Nyeri", 110), ("Embu", 85)],
    "Embu": [("Meru", 75), ("Thika", 85)],
    "Meru": [("Nanyuki", 75)],
    "Nanyuki": [("Nyeri", 55)],
    "Nyeri": [("Nairobi", 115), ("Nyahururu", 80), ("Nanyuki", 55)],
    "Nyahururu": [("Nakuru", 105), ("Eldama Ravine", 70), ("Nyeri", 80)],
    "Eldama Ravine": [("Eldoret", 90), ("Nyahururu", 70)],
    "Nakuru": [("Naivasha", 65), ("Nyahururu", 105), ("Eldama Ravine", 70), ("Eldoret", 155), ("Molo", 50), ("Nyeri", 185)],
    "Naivasha": [("Nakuru", 65), ("Narok", 100), ("Nairobi", 90)],
    "Narok": [("Naivasha", 100), ("Bomet", 85)],
    "Bomet": [("Narok", 85), ("Kericho", 60), ("Kisii", 75)],
    "Kisii": [("Bomet", 75), ("Kisumu", 115)],
    "Kericho": [("Bomet", 60), ("Molo", 65), ("Kisumu", 100)],
    "Molo": [("Nakuru", 50), ("Kericho", 65)],
    "Eldoret": [("Webuye", 75), ("Kapsabet", 40), ("Eldama Ravine", 90), ("Nakuru", 155)],
    "Kapsabet": [("Eldoret", 40), ("Kisumu", 120)],
    "Webuye": [("Eldoret", 75), ("Kakamega", 40)],
    "Kakamega": [("Webuye", 40), ("Kisumu", 55)],
    "Kisumu": [("Kakamega", 55), ("Kapsabet", 120), ("Kericho", 100), ("Kisii", 115)]
}
```

```python
import heapq

def uniform_cost_search(graph, start, goal):
    # Priority queue stores: (total_cost, current_node, path_so_far)
    frontier = [(0, start, [start])]
    visited = set()
    min_costs = {start: 0}
    step = 1

    print(f"Uniform Cost Search from {start} to {goal}\n")

    while frontier:
        cost, current, path = heapq.heappop(frontier)

        if current in visited:
            continue

        print(f"Step {step}:")
        print(f"  Expanding: {current} (Cost: {cost} km)")
        print(f"  Path: {' -> '.join(path)}")

        # Goal test happens when the node is popped
        if current == goal:
            print(f"\nSuccess! Optimal path found:")
            print(f"Path: {' -> '.join(path)}")
            print(f"Total Cost: {cost} km")
            return path, cost

        visited.add(current)

        for neighbor, weight in graph.get(current, []):
            new_cost = cost + weight
            if neighbor not in visited and (neighbor not in min_costs or new_cost < min_costs[neighbor]):
                min_costs[neighbor] = new_cost
                heapq.heappush(frontier, (new_cost, neighbor, path + [neighbor]))

        frontier_preview = [(n, c) for c, n, _ in sorted(frontier)]
        print(f"  Visited: {sorted(list(visited))}")
        print(f"  Frontier: {frontier_preview}\n")
        step += 1

    print("No path found to the destination.")
    return None, float("inf")

if __name__ == "__main__":
    uniform_cost_search(highway_graph, "Nairobi", "Kisumu")
```
